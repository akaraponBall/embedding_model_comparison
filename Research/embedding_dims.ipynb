```python
import numpy as np
from tqdm import tqdm
from langchain_ollama import OllamaEmbeddings
```

```python
MODELS = [
    "bge-m3",
    "qwen3-embedding:0.6b",
    "embeddinggemma",
    "bge-large",
    "nomic-embed-text",
    "mxbai-embed-large"
]

sentence = 'Maholan (มโหฬาร) in Thai means gigantic. We perceive big data as an invaluable resource shaping the future. \
The data landscape of tomorrow will be characterized by its immense scale, rapid evolution, and inherent complexity. \
This demands meticulous capturing, systematic organization, strategic monetization, and the artful refinement of data.'
```

```python
for model in MODELS:
    embedder = OllamaEmbeddings(model=model, keep_alive=0)
    embeded_vec = embedder.embed_documents([sentence])
    embedding_dims = np.array(embeded_vec).squeeze().shape
    print(f'model: {model} -- Embedding Dimension = {embedding_dims}')
```

```python
import requests

url = "http://localhost:11434/api/embed"   # Native API endpoint
payload = {
    "model": "qwen3-embedding:0.6b",
    "input": ["hello world"],
    "dimensions": 512  # pick any value between 32 and 4096
}

response = requests.post(url, json=payload)
data = response.json()

print("Embedding length:", len(data["embeddings"][0]))
print("First 10 values:", data["embeddings"][0][:10])
```
